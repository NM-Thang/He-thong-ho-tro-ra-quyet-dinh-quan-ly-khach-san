```python
import pyodbc
import pandas as pd
```

```python
def create_connection():
    # Thông tin kết nối SQL Server
    server = 'ZOHATEA'  # Tên server từ SSMS
    database = 'Hotel'  # Đã sửa, bỏ dấu ;
    username = 'sa'
    password = 'ptit'

    # Kết nối SQL Server
    try:
        conn = pyodbc.connect(
            f"DRIVER={{ODBC Driver 17 for SQL Server}};"  # Sử dụng driver có sẵn
            f"SERVER={server};"
            f"DATABASE={database};"
            f"UID={username};"
            f"PWD={password}"
        )
        cursor = conn.cursor()
        print("Kết nối thành công!")
    except pyodbc.Error as ex:
        print(f"Lỗi kết nối: {ex}")
    return conn, cursor
```

# 1. ViTri

```python
df_vitri = pd.read_csv('ViTri.csv')

# chen dư liệu vào bảng ViTri
conn, cursor = create_connection()
try:
    cursor.executemany('''
    INSERT INTO ViTri (MaThanhPho, TenThanhPho, VungMien)
    VALUES (?, ?, ?)
    ''', df_vitri.values.tolist())
    conn.commit()
    print(f"ViTri: {len(df_vitri)} ")
except pyodbc.Error as ex:
    print(f"Lỗi khi chèn dữ liệu: {ex}")
    conn.rollback()
# Đóng kết nối  
conn.close()
```
