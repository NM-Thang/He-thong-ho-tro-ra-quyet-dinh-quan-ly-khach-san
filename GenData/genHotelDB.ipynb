# 4. Phong

```python
import random
```

```python
# Tỉ lệ phân bổ loại phòng (%)
room_type_distribution = {
    "Standard Room": 30,
    "Deluxe Room": 20,
    "Superior Room": 12,
    "Executive Room": 5,
    "Junior Suite": 5,
    "Suite": 5,
    "Presidential Suite": 1,
    "Family Room": 7,
    "Connecting Room": 5,
    "Twin Room": 4,
    "Double Room": 5,
    "Single Room": 2,
    "Studio Room": 2,
    "Accessible Room": 2,
    "Penthouse Suite": 1
}

# Đơn giá mỗi đêm (VND)
room_prices_vnd = {
    "Standard Room": 1200000,
    "Deluxe Room": 1800000,
    "Superior Room": 1500000,
    "Executive Room": 2400000,
    "Junior Suite": 2800000,
    "Suite": 3500000,
    "Presidential Suite": 7200000,
    "Family Room": 2200000,
    "Connecting Room": 2000000,
    "Twin Room": 1700000,
    "Double Room": 1700000,
    "Single Room": 1000000,
    "Studio Room": 1300000,
    "Accessible Room": 1100000,
    "Penthouse Suite": 8000000
}
```

```python
# Giả sử mỗi khách sạn có ngẫu nhiên từ 50–70 phòng
def generate_rooms_for_hotel(hotel_id, total_rooms):
    room_data = []
    room_counter = 1

    # Tính số phòng theo tỉ lệ
    total_ratio = sum(room_type_distribution.values())
    for room_type, ratio in room_type_distribution.items():
        num_rooms = round((ratio / total_ratio) * total_rooms)
        for _ in range(num_rooms):
            ma_phong = f"{hotel_id}_R{str(room_counter).zfill(3)}"
            gia_niem_yet = room_prices_vnd[room_type]
            trang_thai = random.choice(["Trống", "Đang sử dụng", "Đang dọn"])
            room_data.append({
                "MaPhong": ma_phong,
                "MaKhachSan": hotel_id,
                "LoaiPhong": room_type,
                "GiaNiemYet": gia_niem_yet,
                "TrangThai": trang_thai
            })
            room_counter += 1

    return room_data
```

```python
all_rooms = []

for _, row in df_khachsan.iterrows():
    hotel_id = row["hotel_id"]
    total_rooms = random.randint(10, 40)
    hotel_rooms = generate_rooms_for_hotel(hotel_id, total_rooms)
    all_rooms.extend(hotel_rooms)

# Tạo DataFrame
df_phong = pd.DataFrame(all_rooms)
df_phong.to_csv('Hotel/Phong.csv', index=False)
df_phong
```
